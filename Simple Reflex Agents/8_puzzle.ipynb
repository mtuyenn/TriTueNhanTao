```python
import random

class SimpleReflexAgent:
    def __init__(self):
        self.rules = {
            0: ['DOWN', 'RIGHT'],
            1: ['DOWN', 'LEFT', 'RIGHT'],
            2: ['DOWN', 'LEFT'],
            3: ['UP', 'DOWN', 'RIGHT'],
            4: ['UP', 'DOWN', 'LEFT', 'RIGHT'],
            5: ['UP', 'DOWN', 'LEFT'],
            6: ['UP', 'RIGHT'],
            7: ['UP', 'LEFT', 'RIGHT'],
            8: ['UP', 'LEFT']
        }

    def simple_reflex(self, percept):
        """Agent phản xạ đơn giản: Dựa vào vị trí của ô trống (0) để chọn hành động"""
        blank_index = percept.index(0)
        
        
        valid_actions = self.rules[blank_index]
        
        return random.choice(valid_actions) 

def apply_action(state, action):
    """Môi trường cập nhật trạng thái bàn cờ"""
    state_list = list(state)
    blank_index = state_list.index(0)
    
    if action == 'UP': new_index = blank_index - 3
    elif action == 'DOWN': new_index = blank_index + 3
    elif action == 'LEFT': new_index = blank_index - 1
    elif action == 'RIGHT': new_index = blank_index + 1

    state_list[blank_index], state_list[new_index] = state_list[new_index], state_list[blank_index]
    return tuple(state_list)

def print_board(state):
    """In bàn cờ 3x3"""
    for i in range(0, 9, 3):
        print(state[i:i+3])
    print("-" * 15)


def run_until_duplicate():
    # Khởi tạo trạng thái ban đầu
    initial_list = [1, 2, 3, 5, 6, 4, 8, 7, 0]
    random.shuffle(initial_list)
    current_state = tuple(initial_list)

    agent = SimpleReflexAgent()

    # Sử dụng Set để lưu các trạng thái đã đi qua (tốc độ kiểm tra siêu nhanh)
    visited_states = set()
    visited_states.add(current_state)

    print("=== TRẠNG THÁI BAN ĐẦU ===")
    print_board(current_state)

    step = 0
    while True:
        step += 1
        
        # 1. Agent lấy trạng thái hiện tại và đưa ra hành động
        action = agent.simple_reflex(current_state)
        print(f"Bước {step}: Agent đi {action}")
        
        # 2. Môi trường thực hiện hành động
        next_state = apply_action(current_state, action)
        print_board(next_state)

        # 3. KIỂM TRA: Trạng thái này đã từng xuất hiện chưa?
        if next_state in visited_states:
            print("DỪNG CHƯƠNG TRÌNH!")
            print(f"Lý do: Trạng thái ở bước {step} đã bị trùng lặp với quá khứ.")
            break
            
        # Nếu chưa trùng, lưu trạng thái này vào lịch sử và đi tiếp
        visited_states.add(next_state)
        current_state = next_state

if __name__ == "__main__":
    run_until_duplicate()
```
