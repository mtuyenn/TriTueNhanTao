PEAS cho máy hút bụi
P: Làm sạch nhiều nhất, ít tốn điện
E: ma trận m x n biểu thị các số 1, 0 với 1: là sàn bẩn, 0 là sàn sạch
A: các hàm để agent hiểu và hành động
S: cơ chế, hàm để agent đọc dữ liệu

```python
import random
```

```python
class SimpleReflexVacuumAgent:
        def __init__(self, room):
                self.room = room
                self.rows = len(room)
                self.cols = len(room[0])

                # lưu trạng thái trước
                self.previous_states = set()
         # INTERPRET INPUT
        def interpret_input(self, x, y):
                return self.room[x][y]
        
        def possible_moves(self, x, y):
                moves = []

                if x > 0:
                        moves.append('UP')
                        x -= 1
                if x < self.rows - 1:
                        moves.append('DOWN')
                        x += 1
                if y > 0:
                        moves.append('LEFT')
                        y -= 1
                if y < self.cols - 1:
                        moves.append('RIGHT')
                        y += 1

                return moves
        # RULES
        def rules(self, state):

                x, y, status = state

                # Nếu bẩn -> hút bụi
                if status == 1:
                        return 'SUCK'

                # Nếu sạch -> random 
                return random.choice(self.possible_moves(x, y))
        # ACTION
        def action(self, x, y):
                state = (x, y, self.interpret_input(x, y))

                # Nếu lặp trạng thái -> STOP
                if state in self.previous_states:
                        return 'STOP', x, y
                self.previous_states.add(state)
                
                action = self.rules(state)

                if action == 'SUCK':
                        self.room[x][y] = 0 
                
                elif action == 'UP':
                        x -= 1

                elif action == 'DOWN':
                        x += 1

                elif action == 'LEFT':
                        y -= 1

                elif action == 'RIGHT':
                        y += 1
                        
                return action, x, y
```

```python
rooms = [
    [0, 1, 0],
    [1, 0, 1],
    [0, 1, 0]
]

agent = SimpleReflexVacuumAgent(rooms)

x = 0
y = 0
step = 0

while True:

    step += 1

    action, x, y = agent.action(x, y)

    print(f'Step {step}: Action = {action}, Position = ({x}, {y})')

    for row in rooms:
        print(row)

    print()

    if action == 'STOP':
        print('Agent stopped because state repeated.')
        break
```
