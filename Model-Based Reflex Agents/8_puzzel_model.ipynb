```python
import random
```

```python
class ModelBasedReflexAgent:
    def __init__(self):
        self.model = [1, 2, 3, 4, 5, 6, 7, 8, 0]
        self.actions = ['up', 'down', 'left', 'right']

    def pass_actions(self):
        return self.actions

    def update_model(self, percept):
        self.model = list(percept)

    def rules(self, state, visited_states):
        index = state.index(0)
        possible_actions = []

        actions = []

        # kiểm tra các hướng có thể đi
        if index not in [0, 1, 2]:
            actions.append('up')

        if index not in [6, 7, 8]:
            actions.append('down')

        if index not in [0, 3, 6]:
            actions.append('left')

        if index not in [2, 5, 8]:
            actions.append('right')

        for action in actions:
            new_state = self.action_result(state, action)
            if new_state not in visited_states:
                possible_actions.append(action)
                    
        return random.choice(possible_actions) if possible_actions else None

    def action_result(self, state, action):
        state_list = list(state)
        index = state_list.index(0)

        if action == 'up':
            new_index = index - 3

        elif action == 'down':
            new_index = index + 3

        elif action == 'left':
            new_index = index - 1

        elif action == 'right':
            new_index = index + 1

        # đổi chỗ
        state_list[index], state_list[new_index] = (
            state_list[new_index],
            state_list[index]
        )

        return tuple(state_list)

    def print_board(self, state):
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print()

    def run_agent(self, initial_state):

        current_state = tuple(initial_state)

        visited_states = set()
        visited_states.add(current_state)

        print("Trạng thái ban đầu:")
        self.print_board(current_state)

        step = 0

        while True:

            action = self.rules(current_state, visited_states)

            if action is None:
                print("Không còn hành động.")
                break

            new_state = self.action_result(current_state, action)

            if new_state in visited_states:
                print("Trạng thái đã được thực hiện trước đó.")
                break

            visited_states.add(new_state)

            current_state = new_state

            step += 1

            print(f"Bước {step}: {action}")
            self.print_board(current_state)


if __name__ == "__main__":

    run_agent = ModelBasedReflexAgent()

    initial_state = (1, 2, 3,
                     6, 5, 4,
                     8, 7, 0)

    run_agent.run_agent(initial_state)
```
