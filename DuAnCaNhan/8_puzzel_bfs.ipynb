```python
from collections import deque
```

```python
goal_state = (
    (1, 2, 3),
    (4, 5, 6),
    (7, 8, 0)
)
```

```python
# In ma trận
def print_state(state):
    for row in state:
        print(row)
    print()
```

```python
# Tìm vị trí ô trống
def find_zero(state):
    for i in range(3):
        for j in range(3):
            if state[i][j] == 0:
                return i, j
```

```python
def extract_state(state):
    return tuple(tuple(row) for row in state)
```

```python
def expand_state(state):
    x, y = find_zero(state)
    children = []
    
    directions = [
        (-1, 0, "UP"),
        (1, 0, "DOWN"),
        (0, -1, "LEFT"),
        (0, 1, "RIGHT")
    ]

    for dx, dy, action in directions:
        nx, ny = x + dx, y + dy
        if 0 <= nx < 3 and 0 <= ny < 3:
            new_state = [list(row) for row in state]
            new_state[x][y], new_state[nx][ny] = (
                new_state[nx][ny],
                new_state[x][y]
            )
            children.append((extract_state(new_state), action))
    return children
```

```python
# BFS
def bfs(initial_state):
    frontier = deque()
    #them node dau tien vao hang doi
    frontier.append((extract_state(initial_state), []))
    visited = set()
    visited.add(extract_state(initial_state))

    while frontier:
        current_state, path = frontier.popleft()

        if current_state == goal_state:
            return path, current_state 
        
        children = expand_state(current_state)
        
        for child_state, action in children:
            if child_state not in visited:
                visited.add(child_state)
                frontier.append((child_state, path + [action]))

    return None
```

```python
initial_state = (
    (1, 2, 3),
    (4, 0, 6),
    (7, 5, 8)
)

print("Ban Đầu:")
print_state(initial_state)

result = bfs(initial_state)

if result:

    path, final_state = result

    print("Giải pháp tìm được!")
    print("Bước:", len(path))
    print("Path:", path)

    print("\nGoal State:")
    print_state(final_state)

else:
    print("Không tìm thấy giải pháp!")
```
